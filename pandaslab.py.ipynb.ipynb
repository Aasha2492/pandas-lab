```python
# Step 2: Check for missing values
print("Missing values per column:")
print("-" * 30)
missing = df.isnull().sum()
print(missing)
print("-" * 30)
print(f"Total missing values: {missing.sum()}")
```

```python
# Step 4: Group by Pclass and find average of all numeric columns
# Pclass (1st, 2nd, 3rd) is the category column in this Titanic dataset

numeric_cols = ['Age', 'Fare', 'SibSp', 'Parch', 'Survived']

print("Average of numeric columns grouped by Pclass (Passenger Class):")
print()
grouped_avg = df.groupby('Pclass')[numeric_cols].mean().round(2)
print(grouped_avg)
```
